```python
from datetime import datetime, timezone

def to_utc(timestamps):
    """Convert Unix timestamps to UTC datetime objects."""
    return [datetime.fromtimestamp(ts, tz=timezone.utc) for ts in timestamps]

# Or if you want formatted strings:
def to_utc_str(timestamps, fmt="%Y-%m-%d %H:%M:%S UTC"):
    """Convert Unix timestamps to formatted UTC strings."""
    return [datetime.fromtimestamp(ts, tz=timezone.utc).strftime(fmt) for ts in timestamps]
```
