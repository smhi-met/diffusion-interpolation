 # Rechunk
 for small dataset where loading 2.5 mkb files is so problematic

```python
import xarray as xr
```

```python
source_zarr_path = "/ec/res4/hpcperm/smcd/data/aifs-meps-2.5km-2020-2023-1h-v2_SMHI_subdomain_reduced.zarr" # Atos 
target_zarr_path = "/ec/res4/hpcperm/smcd/data/aifs-meps-2.5km-2020-2023-1h-v2_SMHI_subdomain_reduced_no_time_chunks.zarr" 
```

```python
import xarray as xr

# 1. Open your existing "slow" dataset
ds = xr.open_zarr(source_zarr_path, zarr_format = 2)

# 2. Define the new chunking strategy
# Setting time to -1 tells Xarray to make it the full length of the dimension
new_chunks = {
    'time': 6,           # All time steps in one chunk
    'variable': -1,      # Keep variables separate (optional)
    'ensemble': 1,
    'cell': 65536         # Keep the spatial block as is
}

# 3. Apply and Save
# 'consolidated=True' is vital for HPC performance
ds_rechunked = ds.chunk(new_chunks)
```

```python
ds_rechunked.to_zarr(target_zarr_path, consolidated=True, mode='w', zarr_format = 2)
```
