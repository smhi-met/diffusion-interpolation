```python
import xarray as xr
import dask

from dask.diagnostics import ProgressBar
```

```python
source_zarr_path = "/project/home/p200177/u101329/DE371_bis/MEPS_subdomain/meps-2p5km-2020-2025-1h-v2_subdomain_subvars.zarr" # Mlx 
target_zarr_path = "/project/home/p200177/u101329/DE371_bis/MEPS_subdomain/meps-2p5km-2020-2025-1h-v2_subdomain_subvars_rech.zarr" 
target_zarr_path = "/mnt/tier1/project/p200177/DE_371_bis//meps-2p5km-2020-2025-1h-v2_subdomain_subvars_rech.zarr" 
```

```python

# 1. Open your existing "slow" dataset
ds = xr.open_zarr(source_zarr_path, zarr_format = 2)

 
new_chunks = {
    'time': 6,
    'variable': -1,  #full dimensions all 
    'ensemble': -1,
    'cell': -1
}
ds_rechunked = ds.chunk(new_chunks)
ds_rechunked.data.encoding = {}
```

```python
with ProgressBar():
    ds_rechunked.to_zarr(
    target_zarr_path, 
    consolidated=True, 
    mode='w', 
    zarr_format=2,
    align_chunks=True 
)
```
